```python
data.drop('trip_id',    axis=1, inplace=True)
data.drop('driver_id',  axis=1, inplace=True)
data.drop('vehicle_id', axis=1, inplace=True)
print("ID columns dropped. Remaining:", data.shape[1], "columns")
```

```python
data['timestamp']   = pd.to_datetime(data['timestamp'])
data['hour']        = data['timestamp'].dt.hour
data['day_of_week'] = data['timestamp'].dt.dayofweek
data['is_weekend']  = (data['day_of_week'] >= 5).astype(int)
data['is_night']    = ((data['hour'] >= 22) | (data['hour'] <= 5)).astype(int)
data.drop('timestamp', axis=1, inplace=True)
print("Timestamp parsed. New features: hour, day_of_week, is_weekend, is_night")
```

```python
data.head()
```

```python
data.info()
```

```python
def fillnaMean(cols):
    for i in cols:
        data[i] = data[i].fillna(data[i].mean())

columns = ['speed','acceleration','heading','trip_duration','trip_distance',
           'fuel_consumption','rpm','lane_deviation','route_deviation_score',
           'acceleration_variation','behavioral_consistency_index',
           'latitude','longitude']
fillnaMean(columns)
print("Numerical columns filled with mean.")
```

```python
def fillnaMode(cols):
    for i in cols:
        data[i] = data[i].fillna(data[i].mode()[0])

columns = ['steering_angle','brake_usage','stop_events',
           'geofencing_violation','anomalous_event','route_anomaly',
           'weather_conditions','road_type','traffic_condition',
           'hour','day_of_week','is_weekend','is_night']
fillnaMode(columns)
print("Categorical/integer columns filled with mode.")
```

```python
data.info()
```

```python
def cap_outliers_iqr(cols):
    for col in cols:
        Q1    = data[col].quantile(0.25)
        Q3    = data[col].quantile(0.75)
        IQR   = Q3 - Q1
        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR
        n_out = ((data[col] < lower) | (data[col] > upper)).sum()
        data[col] = data[col].clip(lower=lower, upper=upper)
        if n_out > 0:
            print(f"  '{col}': {n_out} outliers capped  [{lower:.2f}, {upper:.2f}]")

columns = ['speed','acceleration','rpm','fuel_consumption','trip_duration',
           'trip_distance','lane_deviation','route_deviation_score',
           'acceleration_variation','behavioral_consistency_index','brake_usage']
cap_outliers_iqr(columns)
print("Outlier capping complete.")
```

```python
def changetoInt64(cols):
    for i in cols:
        data[i] = data[i].astype('int64')

columns = ['steering_angle','brake_usage','stop_events','geofencing_violation',
           'anomalous_event','route_anomaly','hour','day_of_week','is_weekend','is_night']
changetoInt64(columns)
print("Integer type conversion done.")
```

## Creating the Categorical Target: `risk_level`

| Risk Level | Condition |
|---|---|
| **High** | `route_anomaly=1` OR `anomalous_event=1` OR `geofencing_violation=1` OR (very high speed AND low consistency) |
| **Medium** | Above-average speed OR high lane deviation OR high route/acceleration deviation |
| **Low** | Normal safe driving |

```python
data['risk_level'] = 'Low'

medium_mask = (
    (data['speed']                  > data['speed'].quantile(0.75)) |
    (data['lane_deviation']         > data['lane_deviation'].quantile(0.75)) |
    (data['route_deviation_score']  > data['route_deviation_score'].quantile(0.75)) |
    (data['acceleration_variation'] > data['acceleration_variation'].quantile(0.75)) |
    (data['brake_usage']            > data['brake_usage'].quantile(0.75))
)
data.loc[medium_mask, 'risk_level'] = 'Medium'

high_mask = (
    (data['route_anomaly']        == 1) |
    (data['anomalous_event']      == 1) |
    (data['geofencing_violation'] == 1) |
    (
        (data['speed'] > data['speed'].quantile(0.90)) &
        (data['behavioral_consistency_index'] <
         data['behavioral_consistency_index'].quantile(0.10))
    )
)
data.loc[high_mask, 'risk_level'] = 'High'

print("risk_level distribution:")
print(data['risk_level'].value_counts())
```

```python
data.drop('route_anomaly',        axis=1, inplace=True)
data.drop('anomalous_event',      axis=1, inplace=True)
data.drop('geofencing_violation', axis=1, inplace=True)
print("Leakage columns dropped. Columns remaining:", data.shape[1])
```

```python
encoders = {}

def encodeCols(cols):
    for i in cols:
        value = pd.DataFrame({i: data[i].unique()})
        LE = LabelEncoder()
        LE.fit(value)
        data[i] = LE.transform(data[i])
        encoders[i] = LE
        print(f"  Encoded '{i}': {list(LE.classes_)}")

columns = ['weather_conditions','road_type','traffic_condition','risk_level']
encodeCols(columns)

joblib.dump(encoders, 'encoders.pkl')
print("\nEncoders saved to encoders.pkl")
```

## Training Model using Classification : Random Forest

```python
print("-- Training the Model using Random Forest --")

model_rf = RandomForestClassifier(
    n_estimators = 100,   # 100 trees
    random_state = 42,
    n_jobs       = -1     # use all CPU cores for speed
)
model_rf.fit(train_X, train_y)

print(model_rf)
print("Done!")
```

```python
import pandas as pd, numpy as np, pickle, joblib

# Load Model & Encoders
model    = pickle.load(open('model_rf.pkl', 'rb'))
encoders = joblib.load('encoders.pkl')


def encode_cols(le, value):
    if value in le.classes_:
        return le.transform([value])[0]
    else:
        return -1

print("Enter User Input:")

user_input = {
    "latitude"                    : float(input("Latitude (e.g. 37.77): ")),
    "longitude"                   : float(input("Longitude (e.g. -122.41): ")),
    "speed"                       : float(input("Speed km/h (e.g. 45.0): ")),
    "acceleration"                : float(input("Acceleration m/s2 (e.g. 1.5): ")),
    "steering_angle"              : int(input("Steering Angle -45 to 45 (e.g. 10): ")),
    "heading"                     : float(input("Heading 0-360 (e.g. 180.0): ")),
    "trip_duration"               : float(input("Trip Duration minutes (e.g. 30.0): ")),
    "trip_distance"               : float(input("Trip Distance km (e.g. 15.0): ")),
    "fuel_consumption"            : float(input("Fuel Consumption L/100km (e.g. 8.5): ")),
    "rpm"                         : float(input("RPM (e.g. 2500.0): ")),
    "brake_usage"                 : int(input("Brake Usage count 0-19 (e.g. 3): ")),
    "lane_deviation"              : float(input("Lane Deviation 0.0-1.0 (e.g. 0.2): ")),
    "weather_conditions"          : input("Weather (Sunny / Cloudy / Foggy / Rainy): "),
    "road_type"                   : input("Road Type (Urban / Highway / Rural): "),
    "traffic_condition"           : input("Traffic (Light / Moderate / Heavy): "),
    "stop_events"                 : int(input("Stop Events count (e.g. 2): ")),
    "route_deviation_score"       : float(input("Route Deviation Score 0.0-1.0 (e.g. 0.1): ")),
    "acceleration_variation"      : float(input("Acceleration Variation 0.0-1.0 (e.g. 0.2): ")),
    "behavioral_consistency_index": float(input("Behavioral Consistency Index 0.0-1.0 (e.g. 0.8): ")),
    "hour"                        : int(input("Hour 0-23 (e.g. 14): ")),
    "day_of_week"                 : int(input("Day of Week 0=Mon 6=Sun (e.g. 2): ")),
    "is_weekend"                  : int(input("Is Weekend 0=No 1=Yes: ")),
    "is_night"                    : int(input("Is Night 0=No 1=Yes: ")),
}

# Encode categorical string inputs
for col in ['weather_conditions','road_type','traffic_condition']:
    user_input[col] = encode_cols(encoders[col], user_input[col])


features = [
    'latitude','longitude','speed','acceleration','steering_angle',
    'heading','trip_duration','trip_distance','fuel_consumption','rpm',
    'brake_usage','lane_deviation','weather_conditions','road_type',
    'traffic_condition','stop_events','route_deviation_score',
    'acceleration_variation','behavioral_consistency_index',
    'hour','day_of_week','is_weekend','is_night'
]

X_input    = np.array([[user_input[col] for col in features]])
predict    = model.predict(X_input)
risk_label = encoders['risk_level'].inverse_transform(predict)[0]

print("\nPredicted Driving Risk Level:", risk_label)

advice = {
    'Low'   : 'Safe trip. Keep it up!',
    'Medium': 'Moderate risk. Drive carefully.',
    'High'  : 'HIGH RISK! Take immediate precautions.',
}
print("Advice:", advice.get(risk_label,""))
```
